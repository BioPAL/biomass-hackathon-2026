# Team 11a — BIOMASS soil moisture: pick a tile and run

One panel. **Map**: all 3 903 L1C tiles as one vector layer (black outline = ISMN station inside, blue = stacks found at your point, colour = screening score once `tile_screening.csv` exists; tooltip with stacks, period and numbers); *Map options* can thin the view to station/screening tiles. Layers (top right): OSM, Esri satellite, ESA WorldCover 2021.

**Flow**: click the map → the coordinate is copied → paste it into *paste lat, lon* (or type a *stack key*) → **Find** → choose the stack in the dropdown → **Run** (tick *reference only* for a run without token). The run log unfolds below; when it is done the results appear as tabs: main figure, coherence, quicklooks, key numbers, check figures.

**Extras** (folded): *Candidate windows in the tile* rates all 600 m blocks by land cover and slope (`criteria.json`) and lets you adopt one of the best as the point; *Reference preview* shows all soil-moisture sources before any SAR data are touched.

Paths: token in `~/token.txt` (or `~/my-private-bucket/token.txt`); outputs in `~/data/runs/<stack>/`; ISMN exports from the shared bucket (first code cell).
