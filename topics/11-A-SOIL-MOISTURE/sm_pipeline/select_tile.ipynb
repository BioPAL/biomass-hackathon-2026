# Team 11a - BIOMASS soil moisture: global dashboard

All 3 903 BIOMASS L1C tiles (24 904 stacks) on one map, coloured and filtered by land cover, slope, ISMN stations and acquisition month (information for finding test sites). **Click a tile** → its land cover, stacks and the public BPS quicklooks. **Click a stack** → its footprint (dashed). **Click inside the footprint** to set your own point and choose the **analysis window** (200-2 000 m, default 600 m); without a point the run uses the ISMN stations inside the stack, or the stack centre. **compute full report** runs `run_stack.py` for that stack (several minutes, log below the map) and shows the full report with all figures below; **compute all N stacks at this point** runs every stack that contains the point and passes the filters, one after the other.

Computed points are dots: **yellow** = with in-situ station data (ISMN), **orange** = without in-situ data (model reference only), **grey** = SAR skipped. Click a dot to open its stack and **open full report**.

Paths: token in `~/token.txt` (or `~/my-private-bucket/token.txt`); runs in `~/data/runs/<stack>/` (report: `report.html`); ISMN exports from the shared bucket. The first start builds `dashboard.html` (about 30 s); *rebuild dashboard* re-reads the runs.
